```python
L_coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
L_coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
L_coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]
import matplotlib.pyplot as plt
import numpy as np

def average_UE(notes, coef):
   moy = round(np.average(notes, weights=coef), 3)
   return moy
       
def moy_UE_graph(L1, L2, L3):
   try:
       matiere = [f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)]
       liste_note = []
       
       nb_note = int(input("combien de notes avez vous ? "))
       if nb_note > len(L1):
           print(f"Vous ne pouvez pas avoir plus de {len(L1)} notes")
           nb_note = len(L1)
       
       for i in range(nb_note):
           liste_note.append(float(input(f"Quelle est la note de {matiere[i]} ? ")))
       
       moy_ue1 = average_UE(liste_note, L1[:nb_note])
       moy_ue2 = average_UE(liste_note, L2[:nb_note])
       moy_ue3 = average_UE(liste_note, L3[:nb_note])
       
       UE = [f"UE{a}" for a in range(1, 4)]
       moyennes = [moy_ue1, moy_ue2, moy_ue3]
       
       plt.bar(UE, moyennes)  
       plt.title("Moyenne des UE du S1")
       plt.xlabel("UE")
       plt.ylabel('Notes (sur 20)')
       plt.xticks(rotation=45) 
       plt.tight_layout() 
       plt.show()
       
       print(f"Votre moyenne de l'UE1 est: {moy_ue1:.2f}")
       print(f"Votre moyenne de l'UE2 est: {moy_ue2:.2f}")
       print(f"Votre moyenne de l'UE3 est: {moy_ue3:.2f}")
       
       return [moy_ue1, moy_ue2, moy_ue3]
       
   except ValueError:
       print("Entrée invalide. Veuillez réessayer.")
       return None

result = moy_UE_graph(L_coef1, L_coef2, L_coef3)
```
