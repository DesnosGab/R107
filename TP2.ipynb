```python
Lcoef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
Lcoef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
Lcoef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]
import matplotlib.pyplot as plt 
import numpy as np

def moyUEgraphique(L):
   moyenne=0
   listenote=[]
   coeftot=0
   NBnote=int(input("combien de note avez vous?"))
   for i in range(101,101+NBnote):
       listenote.append(int(input("indiquez vos note dans l'ordre")))
   for k in range(NBnote):
       moyenne+=listenote[k]*L[k]
       coeftot+=L[k]
   moyenne=moyenne/coeftot
   print(moyenne)
   return moyenne


moy1=moyUEgraphique(Lcoef1)
moy2=moyUEgraphique(Lcoef2)
moy3=moyUEgraphique(Lcoef3)

liste_moy=[]
liste_moy.append(moy1)
liste_moy.append(moy2)
liste_moy.append(moy3)
liste_UE=["UE1","UE2","UE3"]

plt.bar(liste_UE,liste_moy)
plt.title("moyenne des notes du semestre 1")
plt.xlabel("UE")
plt.ylabel("moyenne (sur 20)")
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```
